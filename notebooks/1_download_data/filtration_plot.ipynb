```python
import matplotlib
import seaborn as sns
matplotlib.rcParams['pdf.fonttype'] = 42
matplotlib.rcParams['ps.fonttype'] = 42
matplotlib.rcParams['svg.fonttype'] = 'none'
matplotlib.rcParams['font.sans-serif'] = 'Arial'
matplotlib.rcParams['font.family'] = 'sans-serif'
sns.set_style('ticks')
matplotlib.rcParams['text.color'] = '#000000'
matplotlib.rcParams['axes.labelcolor'] = '#000000'
matplotlib.rcParams['xtick.color'] = '#000000'
matplotlib.rcParams['ytick.color'] = '#000000'
```

```python
from plotly import graph_objects as go

# Data
y_labels_process = ["L50/N50", "Contig", "CheckM Completeness Contamination",
                   "Minimum Genome Length", "Maximum Predicted CDS", "GC content",
                   "Bad Downloads", "De-duplicate", "MASH", "All"]

# based on number of strains leftover from various filtration steps in 1a, 1b, 1c, 2a, and 2b
x_values_process = [7114 + 519, 6324 + 515, 5338 + 515, 3094 + 407, 3093 + 407, 3090 + 377, 3014 + 368, 3225, 3107, 3068] 

# Updated color palette with more vibrant colors
colors_process = [
    '#1E3A8A',  # Strong blue
    '#2563EB',  # Royal blue
    '#4F46E5',  # Medium blue
    '#6366F1',  # Periwinkle blue
    '#A3BFFA',  # Soft blue
    '#16A34A',  # Green
    '#10B981',  # Emerald green
    '#D97706',  # Orange
    '#DC2626',  # Red
    '#F87171'   # Light red
]

# Create annotations for labels between blocks
annotations_process = []
for i in range(len(y_labels_process) - 1):
    position = (i + i + 1) / 2
    annotations_process.append({
        'x': max(x_values_process) - 1400,  # Adjust this for label position
        'y': position,
        'xref': 'x',
        'yref': 'y',
        'text': y_labels_process[i],
        'showarrow': False,
        'font': {
            'family': 'Palatino Linotype, serif',
            'size': 16,
            'color': 'black'
        }
    })

# Plotting
fig_process = go.Figure(go.Funnel(
    y=y_labels_process,
    x=x_values_process,
    textposition="inside",  # Text inside each step of the funnel
    marker={"color": colors_process},
    textfont={"family": "Palatino Linotype, serif", "size": 13, "color": "black"}
))

# Layout update
fig_process.update_layout(
    width=900,  # Adjusted for space on the right
    height=800,
    annotations=annotations_process,
    funnelmode='stack',  # Makes the funnel shape sharper
    yaxis=dict(
        tickfont=dict(
            family="Palatino Linotype, serif",
            size=13,
            color="black"
        ),
        showticklabels=False  # Hide original y-axis labels
    ),
    xaxis=dict(
        tick0=0,
        dtick=1000,
        nticks=20
    ), # Set background colors to white
    plot_bgcolor='white',
    paper_bgcolor='white'
)

fig_process.show()
fig_process.write_image("../images/supplemental/filtration_plot_all_strains.svg", scale = 3)
```
